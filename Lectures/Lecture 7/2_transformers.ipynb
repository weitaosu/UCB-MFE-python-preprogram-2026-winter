We'll calculate trailing returns, financial ratios, and sector comparisons, then assemble the numeric and categorical columns for a sklearn model. The examples use `equities_class.duckdb`; the [data guide](../modeling/README.md) describes its fields and known limitations.

## 1. Calculate trailing features on a trading calendar

A trailing 63-session return compares today's price with the price 63 market sessions earlier. It is intended to describe roughly three months of price movement. If a stock is missing some quotes, simply moving back 63 rows in its own table could reach much further into the past.

The code first places AAPL on the benchmark's trading calendar with `reindex(calendar)`. Each row then represents a market session, including sessions when AAPL's quote is missing. A missing quote stays missing: filling it with yesterday's price would manufacture a zero return and could understate volatility.

Our signals are calculated after the close, with orders filled at the next session's close. Today's closing price is therefore available for today's features. If we wanted to submit the order before today's close, we would need to build the features from earlier prices instead.

We calculate AAPL's features explicitly, then compare its momentum column with `price_features`, which performs the calculation separately for every stock. Keeping stocks separate prevents a rolling window from combining the end of one ticker's history with the start of another's.

```python
aapl = prices.loc[prices.symbol.eq('AAPL')].set_index('date').reindex(calendar)
px = aapl.adj_close
trailing = pd.DataFrame(index=calendar)
trailing['momentum_63'] = px.pct_change(63, fill_method=None)
trailing['momentum_252_21'] = px.shift(21) / px.shift(252) - 1
trailing['volatility_63'] = px.pct_change(fill_method=None).rolling(63).std() * np.sqrt(252)
trailing['drawdown_252'] = px / px.rolling(252).max() - 1
display(trailing.loc['2014-01-02':'2014-01-09'])
price_panel = price_features(prices, calendar)
pd.testing.assert_series_equal(
    price_panel.loc[price_panel.symbol.eq('AAPL')].set_index('date').momentum_63,
    trailing.momentum_63, check_names=False)
```

The features describe different aspects of the price path. `momentum_63` asks whether the stock has risen over the recent quarter. `momentum_252_21` compares the price 21 sessions ago with the price 252 sessions ago, leaving out the most recent month. A stock that rose over the year but fell sharply last month can therefore have positive longer-term momentum and negative recent momentum.

`volatility_63` measures variation in daily returns, while `drawdown_252` measures how far today's price is below the highest close in the past 252 sessions. In the chart, a drawdown close to zero means the stock is near that trailing high. These features let a model distinguish price trends from the risk taken along the way.

Compute this history before passing rows to sklearn. When predicting one stock-date, the model needs the already calculated trailing feature, not just that day's raw price. A transformer that calculates a rolling window using only the validation batch would lose the earlier history needed for its first rows. After historical features are built, clipping and scaling can be fitted on the training subset.

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
trailing.loc['2012':'2014', ['momentum_63', 'momentum_252_21']].plot(ax=axes[0])
trailing.loc['2012':'2014', ['volatility_63', 'drawdown_252']].plot(ax=axes[1])
axes[0].set_ylabel('Trailing return')
axes[1].set_ylabel('Annualized volatility / drawdown')
plt.tight_layout()
```

## 2. When could an investor have known a financial result?

A statement's fiscal date identifies the period being reported. Its filing date tells us when that report was published. Joining financial results to daily prices using the fiscal date would let the model use information before its publication.

For example, the extract contains Apple's quarter ending **December 28, 2013**, with a filing date of **January 28, 2014**. Its net margin is about 22.70%. A forecast made in early January could not use that number. It would still need to use an earlier available statement.

The source gives dates but no filing times. We therefore make a statement usable from **the next calendar day**: January 29 in this example. This avoids assuming a filing was available before the market close on its publication date. A Friday filing becomes available on Saturday, so its first possible after-close signal is normally Monday, followed by a Tuesday entry.

Some source records have no filing date or a filing date on or before the fiscal period end. We cannot establish reliable availability for those records, so the code counts and excludes them before building the features.

```python
date_issue = fundamentals.filing_date.isna() | fundamentals.filing_date.le(fundamentals.fiscal_date)
display(pd.Series({'statement_rows': len(fundamentals), 'untrusted_filing_dates': int(date_issue.sum()),
                   'unknown_currency': int(fundamentals.reported_currency.isna().sum())}))
display(fundamentals.loc[date_issue, ['symbol', 'fiscal_date', 'filing_date']].head())
events = financial_events(fundamentals)
display(events.loc[events.symbol.eq('AAPL')].tail(4))
```

### Turn statement values into comparable ratios

Raw net income mostly distinguishes large companies from small ones. Dividing by revenue gives net margin, which asks how much profit the company earned per dollar of sales. Similarly, dividing operating income by assets relates operating profit to the size of the asset base.

The example below uses one AAPL statement. These are **quarterly** ratios: a quarter's income or cash flow divided by that statement's revenue or assets. Income/assets is not annualized here. A trailing-twelve-month flow would require four distinct, comparable quarters whose values were all available at the signal date.

| Feature | Calculation | Possible interpretation |
|---|---|---|
| Net margin | net income / revenue | Profit retained from each dollar of sales |
| Operating ROA | operating income / total assets | Operating profit relative to the asset base |
| Debt/assets | total debt / total assets | Balance-sheet leverage |
| Cash/assets | cash / total assets | Cash available relative to company assets |
| FCF/assets | free cash flow / total assets | Cash generation relative to the asset base |
| Accruals/assets | (net income − operating cash flow) / assets | The extent to which reported profit exceeds operating cash flow |

Use positive denominators and leave a ratio missing when its inputs are unusable. Replacing a missing denominator with zero would produce an infinite ratio, while replacing the finished ratio with zero would assert an economic fact we do not know.

Ratios within the same statement avoid mixing market prices in dollars with accounts reported in another currency, although vendor units must still be consistent. Interpret ratios by sector too: debt/assets for a bank describes a different business model from debt/assets for an industrial company.

```python
statement = fundamentals.loc[fundamentals.symbol.eq('AAPL') & ~date_issue].sort_values('fiscal_date').iloc[-1]
worked_ratios = pd.Series({
    'net_margin': statement.net_income / statement.revenue,
    'operating_roa': statement.operating_income / statement.total_assets,
    'debt_to_assets': statement.total_debt / statement.total_assets,
    'accruals_to_assets': (statement.net_income - statement.operating_cash_flow) / statement.total_assets,
})
display(worked_ratios)
# The helper applies positive-denominator guards to every record, not just this example.
print('Ratio units: decimals; net margin 0.20 means 20%, not 0.20%.')
```

## 3. Attach the latest available statement to each daily row

A stock has a price on most trading days but only a few new financial statements each year. For each signal date, we want the latest statement already available at that time. An exact-date merge would leave almost every daily row unmatched.

`merge_asof` solves this by looking backward from each signal date to the nearest `available_date` at or before it. Sort both inputs first and perform the join separately for each stock. The AAPL table below should keep the September 2013 quarter's net margin, about 20.05%, through January 28. It should switch to the December quarter's 22.70% margin on January 29. Inspect `fiscal_date`, `filing_date`, and `available_date` together to see why the switch happens there.

A forward join would do the opposite: it could assign the January 29 statement to earlier January signals. Those forecasts would then contain unpublished results.

The helper also handles two less obvious cases. A late filing for an older fiscal period must not replace a newer period already available. And if the latest known period is more than 550 calendar days old, its ratios are blanked: repeatedly carrying an old value would otherwise make stale accounts look current. The dates remain in the table so we can inspect the age and missingness.

```python
aapl_history = price_panel.loc[price_panel.symbol.eq('AAPL')].sort_values('date')
aapl_events = events.loc[events.symbol.eq('AAPL')].drop(columns='symbol').sort_values('available_date')
joined_example = pd.merge_asof(aapl_history, aapl_events, left_on='date', right_on='available_date', direction='backward')
display(joined_example.loc[joined_example.date.between('2014-01-22', '2014-02-05'),
                           ['date', 'fiscal_date', 'filing_date', 'available_date', 'net_margin']])
features = join_financials(price_panel, events, universe)
assert (features.available_date.isna() | features.available_date.le(features.date)).all()
display(features[NUMERIC_FEATURES].isna().mean().sort_values(ascending=False).rename('missing_fraction'))
```

### Use the stocks that belonged to the universe at the time

The full database holds **3,849 distinct securities**, but they were not all candidates on every date. It contains the union of quarterly selections from 2000-Q1 through 2026-Q3. Each quarter selects up to 2,000 stocks using the preceding market close and quarterly shares recorded as available by then. Some early quarters have fewer qualifying records.

Using all 3,849 names from the beginning would let companies selected much later enter earlier portfolios. Filtering by today's active status would introduce another bias by deleting companies that subsequently disappeared. Instead, `universe_membership` records each quarter's selected stocks, including names now marked inactive. `universe` is simply the full list of stored stocks and their snapshot metadata.

Membership is matched to the **entry date**. A June 30 after-close signal leads to a July 1 purchase and therefore uses Q3 membership, which was selected at the June 30 close. In the table below, inspect the corresponding March 31/April 1 change for AAPL.

Build trailing features before filtering membership, because a newly selected stock may already have useful price history. Keep an exiting stock's prices too: leaving the universe prevents new purchases but does not erase or sell existing holdings. `apply_membership` adds eligibility columns while retaining the history rows.

```python
membership = load_membership()
features = apply_membership(features, membership, calendar)
display(membership.loc[membership.symbol.eq('AAPL') & membership.quarter_start.between('2014-01-01','2014-07-01')])
display(features.loc[features.symbol.eq('AAPL') & features.date.between('2014-03-28','2014-04-02'),
                     ['date','symbol','membership_quarter','universe_selection_date','universe_rank','is_universe_member','eligible']])
assert features.loc[features.eligible, 'universe_selection_date'].le(features.loc[features.eligible, 'date']).all()
print('Membership limits entry candidates; all', len(features), 'history rows remain available.')
```

### Measure performance relative to sector peers

A stock can have positive momentum simply because its whole sector rose. Subtracting its sector's same-date median helps distinguish that common move from the stock's relative performance.

On June 30, 2014, the class extract's three eligible Technology stocks have these approximate 63-session returns:

| Stock | Trailing return | Return minus sector median |
|---|---:|---:|
| AAPL | 21.83% | 19.40 percentage points |
| MSFT | 2.43% | 0.00 percentage points |
| IBM | −5.28% | −7.71 percentage points |

MSFT supplies the median. Apple's positive relative value says it outperformed these peers over the lookback; it is not a forecast of another 19.40% gain. The code also ranks operating ROA within each date and sector, where a rank of 1 identifies the highest value in that group.

Use only stocks eligible on that signal date. A full-history sector average would include later information. We require at least three observed values for either peer statistic and leave smaller groups missing. If you use peer features in your strategy, decide how to handle small or missing groups. Compute the peer features for the full eligible cross-section before passing individual rows to a model.

```python
relative = features.loc[features.eligible].copy()
sector_groups = relative.groupby(['date', 'sector'])
peer_count = sector_groups.momentum_63.transform('count')
relative['sector_relative_momentum'] = (relative.momentum_63 - sector_groups.momentum_63.transform('median')).where(peer_count >= 3)
profit_count = sector_groups.operating_roa.transform('count')
relative['sector_profitability_rank'] = sector_groups.operating_roa.rank(pct=True).where(profit_count >= 3)
display(relative.loc[relative.date.eq('2014-06-30'),
                     ['symbol', 'sector', 'momentum_63', 'sector_relative_momentum', 'sector_profitability_rank']])
print('A rank of 1 is highest within this date/sector; it does not mean a 100% expected return.')
```

The date join still has a source-data limitation. The database keeps the latest vendor values for each fiscal period, so a later restatement may have overwritten the originally published number. Selecting the right filing date cannot recover a value that is no longer stored. Sector labels also come from the source snapshot and may differ from historical classifications.

## 4. Calculate the return the strategy would actually hold

For the AAPL signal on **August 31, 2012**, the next session in the supplied calendar is **September 4**. That is our entry. Adding 90 calendar days gives **December 3**, which has an exit quote. The target is December 3's adjusted close divided by September 4's adjusted close, minus one.

Moving forward 90 trading rows from the same entry instead lands on **January 15, 2013**. That includes several additional weeks of price movement and answers a different forecasting question. This is why `shift(-90)` is not the label for a 90-calendar-day holding rule.

The next cell compares those two returns using the actual prices, then checks the helper's label against the calendar-day calculation. The target is a gross adjusted-price return. Fees are calculated from purchases and sales when we simulate the portfolio.

```python
signal_date = pd.Timestamp('2012-08-31')
entry_date = calendar[calendar.searchsorted(signal_date, side='right')]
due_date = entry_date + pd.Timedelta(days=90)
stock_prices = aapl.adj_close.dropna()
exit_date = stock_prices.index[stock_prices.index.searchsorted(due_date)]
calendar_target = stock_prices.loc[exit_date] / stock_prices.loc[entry_date] - 1
wrong_exit = calendar[calendar.get_loc(entry_date) + 90]
display(pd.DataFrame({'entry': [entry_date, entry_date], 'exit': [exit_date, wrong_exit],
                      'return': [calendar_target, stock_prices.loc[wrong_exit] / stock_prices.loc[entry_date] - 1]},
                     index=['90 calendar days', '90 market sessions']))
panel = make_labels(features, prices, calendar)
row = panel.loc[panel.symbol.eq('AAPL') & panel.date.eq(signal_date)].iloc[0]
assert row.label_end == exit_date and np.isclose(row.target, calendar_target)
display(panel.tail(3)[['date', 'symbol', 'entry_date', 'label_end', 'target']])
```

For signals near the end of the database, the scheduled exit may be beyond the available history. We know the features but cannot yet observe the completed return. This is called **right-censoring**. The label helper leaves those targets missing; it also leaves a target missing if no exit quote arrives within five market sessions of maturity.

A model needs observed targets when fitting or when measuring prediction error. It does not need them to make a forecast. Keep these rows in the prediction set: deleting a stock because its future quote later disappears would use hindsight to change the investment candidates.

## 5. Give sklearn a usable numeric input table

The joined table contains numeric ratios, missing values, and sector names. Ridge needs numeric inputs with no missing values, so different columns require different preparation. `ColumnTransformer` lets us specify a numeric branch and a sector branch, then joins their outputs into the matrix passed to Ridge.

In the numeric branch, clipping is followed by `SimpleImputer(strategy='median', add_indicator=True)`. The imputer learns each column's training median and uses it to fill missing values in both training and validation. For columns that were missing during training, it also creates a zero/one indicator. This lets the model distinguish “the reported ratio was near the median” from “the ratio was unknown and we filled it.” Scaling is applied after these steps.

`keep_empty_features=True` keeps columns that were entirely missing in training, so they do not disappear from the output. It cannot give the model evidence about a ratio never observed during fit. Also, indicators are created from training missingness: a column that first becomes missing during prediction does not automatically gain a new indicator.

In the sector branch, `OneHotEncoder` creates a column for each learned category, such as `sector_Technology`, with 1 for membership and 0 otherwise. Coding sectors as integers would invent an ordering and numeric distances between them. With `handle_unknown='ignore'`, a sector absent from training produces zeros across this branch because there is no learned column for it.

The code passes an explicit `FEATURES` list and sets `remainder='drop'`. Thus identifiers and future information such as `target` and `label_end` cannot slip into the model just because they exist in the table. Inspect `transformed.T` and `get_feature_names_out()` to see how a few original rows became the final model matrix.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
# QuantileClipper is imported from course_data: its implementation was taught in Notebook 1.
numeric = Pipeline([('clip', QuantileClipper()),
                    ('impute', SimpleImputer(strategy='median', add_indicator=True, keep_empty_features=True)),
                    ('scale', StandardScaler())])
preprocessor = ColumnTransformer([
    ('numeric', numeric, NUMERIC_FEATURES),
    ('sector', OneHotEncoder(handle_unknown='ignore', sparse_output=False), ['sector'])
], remainder='drop')
pipe = Pipeline([
    ('preprocess', preprocessor), 
    ('model', Ridge(alpha=100))
])
train = panel.loc[panel.eligible & panel.date.ge('2010-01-01') & panel.label_end.lt('2013-01-01')]
valid = panel.loc[panel.eligible & panel.date.between('2013-01-01', '2013-12-31') & panel.target.notna()]
pipe.fit(train[FEATURES], train.target)
names = pipe['preprocess'].get_feature_names_out()
transformed = pd.DataFrame(pipe['preprocess'].transform(valid[FEATURES].head()),
                           columns=names, index=valid.index[:5])
display(transformed.T)
print('Raw columns:', len(FEATURES), '; transformed columns:', len(names))
imputer = pipe['preprocess'].named_transformers_['numeric']['impute']
print('Columns with training missingness indicators:',
      np.asarray(NUMERIC_FEATURES)[imputer.indicator_.features_].tolist())
assert np.isfinite(transformed.to_numpy()).all()
assert np.isfinite(pipe.predict(valid[FEATURES])).all()
```

The baseline's `FEATURES` list excludes the two sector-relative columns calculated above. To test them, add them to the numeric feature list and compare against the baseline using the same validation dates. `make_pipeline` builds the preprocessing shown here for the remaining examples.
