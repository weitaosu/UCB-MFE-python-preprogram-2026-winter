## 1. What are we asking the model to learn?

Each observation describes **one stock after the market closes on one date**. The inputs summarize what we know at that point. The target is the stock's subsequent holding return: buy at the next session's close, hold for 90 calendar days, and sell at the first available close on or after maturity.

sklearn uses `X` for the input table and `y` for the targets. If there are 5,000 observations and three features, `X` has shape `(5000, 3)` and `y` has shape `(5000,)`. Row 0 in both must refer to the same stock and signal date. Sorting `X` without making the same change to `y` could pair Apple's features with another observation's return; matching array lengths would not catch that mistake.

`prepare_panel` assembles the stock-date table from the class database. We start with three inputs: trailing 21-session return, trailing 63-session return, and annualized volatility of 63 daily returns. Selecting both `X` and `y` from the same filtered table keeps their rows aligned.

````python
panel = prepare_panel()
small_features = ['momentum_21', 'momentum_63', 'volatility_63']
display(panel[['date', 'symbol', *small_features, 'target']].tail(8))
display(panel.groupby(panel.date.dt.year).agg(
    observations=('symbol', 'size'), stocks=('symbol', 'nunique'), labeled=('target', 'count')))
print(f'{len(panel):,} rows; {panel.memory_usage(deep=True).sum() / 1e6:.1f} MB in pandas')
````

The 30 class stocks were selected using information recorded as available before 2010, without requiring them to survive or have complete histories afterward. On each signal date, `eligible` checks whether a stock belongs to the universe for the next session's entry and has enough trailing price history.

The AAPL row below shows the trading dates: signal after close, entry at the next session's close, and exit on or after entry + 90 calendar days. `label_end` is the date we can first observe the completed holding return.

````python
example = panel.loc[panel.symbol.eq('AAPL') & panel.date.eq('2012-08-31')]
display(example[['symbol', 'date', 'entry_date', 'scheduled_exit', 'label_end',
                 'entry_price', 'exit_price', 'target']])
assert (example.entry_date > example.date).all()
assert (example.label_end >= example.scheduled_exit).all()
````

We fit using signals from 2010 onward whose holding returns were complete before 2013. Then we predict the returns of the 2013 observations and compare those predictions with the outcomes.

The dates matter because a feature can be known long before its target. For example, the AAPL signal on December 3, 2012 enters on December 4 and exits on March 4, 2013. We could have calculated its features in December, but we could not have used its eventual return to train a January model.

In the masks below, `date < '2013-01-01'` checks when the signal existed, while `label_end < '2013-01-01'` checks when its outcome was known. Both conditions must hold. Validation targets are used afterward to measure errors; they are never passed to `fit`.

````python
train = panel.loc[panel.eligible & panel.date.ge('2010-01-01')
                  & panel.date.lt('2013-01-01') & panel.label_end.lt('2013-01-01')].copy()
valid = panel.loc[panel.eligible & panel.date.between('2013-01-01', '2013-12-31')
                  & panel.target.notna()].copy()
X_train, y_train = train[small_features], train.target
X_valid, y_valid = valid[small_features], valid.target
assert train.label_end.max() < valid.date.min()
assert not set(train.date).intersection(valid.date)
display(pd.DataFrame({'rows': [len(train), len(valid)],
                      'first_signal': [train.date.min(), valid.date.min()],
                      'last_signal': [train.date.max(), valid.date.max()],
                      'last_outcome': [train.label_end.max(), valid.label_end.max()]},
                     index=['training', 'validation']))
print('X is two-dimensional:', X_train.shape, '; y is one-dimensional:', y_train.shape)
````

## 2. Understand the sklearn interfaces

A sklearn **estimator** is an object we configure and then fit to data. What it learns depends on the object: `StandardScaler.fit` learns column means and standard deviations; `Ridge.fit` learns regression coefficients. Both are estimators, even though only Ridge forecasts a return.

After fitting, a **transformer** changes the representation of `X`, while a **predictor** produces an answer for each row. `StandardScaler.transform` returns scaled features that another estimator can use. `Ridge.predict` returns predicted returns. This distinction explains why we can put a transformer before a predictor in a pipeline.

| Role | Main interface | What the call does | Example |
|---|---|---|---|
| Estimator | `fit(X, y=None)` | Learns from training data and returns the fitted object | Learn scaling statistics or regression coefficients |
| Transformer | `transform(X)` | Applies a fitted transformation, preserving row count and order | Scale columns or encode sectors |
| Predictor | `predict(X)` | Returns one prediction per input row | Forecast a holding return or predict a class |
| Classifier probabilities | `predict_proba(X)` when supported | Returns a probability for each class | Estimate the probability of a positive return |
| Model scoring | `score(X, y)` when supported | Compares predictions with known targets using a default metric | Ridge reports R²; a classifier commonly reports accuracy |
| Meta-estimator | Contains other estimators | Coordinates their fitting and use | `Pipeline` chains preprocessing and a model |

These roles overlap. A pipeline containing a scaler and Ridge is itself an estimator and a predictor. It exposes `predict` because its last step can predict. A pipeline whose last step is a transformer can instead expose `transform`. The method table below lets us inspect this directly rather than infer it from the class names.

`get_params` and `set_params` read and change an estimator's **hyperparameters**: choices supplied before fitting, such as Ridge's penalty strength `alpha`. The same methods work for a standalone model and for models inside a pipeline. Optional methods such as `partial_fit` are available only on estimators that support them; having `fit` does not imply that an object can update itself incrementally.

````python
from sklearn.dummy import DummyRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.base import clone
from sklearn.exceptions import NotFittedError

objects = {
    'StandardScaler': StandardScaler(),
    'Ridge': Ridge(alpha=100),
    'LogisticRegression': LogisticRegression(),
    'scale → Ridge': Pipeline([('scale', StandardScaler()), ('model', Ridge(alpha=100))]),
}
methods = ['fit', 'transform', 'predict', 'predict_proba', 'score', 'get_params']
display(pd.DataFrame({name: {method: callable(getattr(obj, method, None))
                            for method in methods}
                      for name, obj in objects.items()}).T)
````

## 3. What changes when we call fit?

Creating `DummyRegressor(strategy='mean')` chooses a rule: always forecast the training mean. It has not received any data yet, so it does not know that mean. Calling `predict` at this point raises `NotFittedError`.

`fit(X_train, y_train)` calculates the mean of `y_train` and stores it as `constant_`. Every later prediction uses that stored value, including predictions on dates whose returns the model has never seen. In the output below, the forecasts are identical across stocks because this baseline ignores their features.

**Hyperparameters** are choices supplied before fitting, such as `strategy='mean'` and Ridge's `alpha`. **Learned parameters**, such as regression coefficients, are estimated from the training data during `fit`. sklearn stores fitted values in attributes such as `constant_` and `coef_`, with a trailing underscore. Its API names `get_params` and `set_params` use “params” for constructor inputs, including hyperparameters; these methods do not retrieve or change the fitted coefficients.

`fit` returns the fitted object itself, rather than its predictions. That is why `model = Ridge().fit(X, y)` works, and why we still call `model.predict(...)` separately.

````python
baseline = DummyRegressor(strategy='mean')
print('Hyperparameters:', baseline.get_params())
print('Has learned constant before fit:', hasattr(baseline, 'constant_'))
try:
    baseline.predict(X_valid)
except NotFittedError:
    print('Expected: prediction requires fitted state.')

assert baseline.fit(X_train, y_train) is baseline
np.testing.assert_allclose(baseline.constant_.item(), y_train.mean())
display(pd.DataFrame({'symbol': valid.symbol.head(), 'signal_date': valid.date.head(),
                      'forecast': baseline.predict(X_valid.head())}))
print('Learned training mean:', baseline.constant_.item())
````

## 4. Scale the inputs, then fit a predictor

Ridge reduces overfitting by penalizing large coefficients. The size of a coefficient depends on the units of its input: storing a 10% return as `10` instead of `0.10` changes the coefficient needed for the same forecast. Scaling gives the penalty a more comparable meaning across columns.

For each feature, `StandardScaler.fit` learns the training mean $\mu$ and standard deviation $s$. `transform` then calculates $z=(x-\mu)/s$. Crucially, it uses those **same training values** for validation rows. If we recomputed them on the validation period, future observations would influence how earlier forecasts in that period were processed.

`fit_transform(X_train)` combines learning those statistics and scaling the training data. For validation, use only `transform(X_valid)`. The table below shows the stored means and scales alongside the means of the transformed data. Training means should be near zero; validation means need not be, because the market and feature distributions can change.

The manually fitted Ridge then receives these scaled columns. Its prediction array has one value per validation row, in the same order as `X_valid`.

````python
scaler = StandardScaler().set_output(transform='pandas')
scaled_train = scaler.fit_transform(X_train)
scaled_valid = scaler.transform(X_valid)
assert scaler.n_features_in_ == len(small_features)
np.testing.assert_array_equal(scaler.feature_names_in_, X_train.columns)
display(pd.DataFrame({'learned_mean_': scaler.mean_, 'learned_scale_': scaler.scale_,
                      'scaled_train_mean': scaled_train.mean(),
                      'scaled_valid_mean': scaled_valid.mean()}, index=small_features))
manual_model = Ridge(alpha=100).fit(scaled_train, y_train)
manual_prediction = manual_model.predict(scaled_valid)
display(pd.Series(manual_model.coef_, index=small_features, name='coefficient'))
print('X shape:', X_valid.shape, '; prediction shape:', manual_prediction.shape)
````

`set_output(transform='pandas')` keeps the scaler's output as a DataFrame with column names. This lets us match each coefficient to its feature. Without it, the usual output is an array, so we need to keep track of column order ourselves.

## 5. Put preprocessing and prediction in one Pipeline

The manual calculation required us to fit the scaler, transform training data, fit Ridge, and remember to use the same scaler on validation data. A `Pipeline` stores these steps together and calls them in order.

In the pipeline below, `scale` and `model` are names we choose for the two steps. `pipe.fit(X_train, y_train)` first runs the scaler's `fit_transform`, then passes the scaled training table and `y_train` to Ridge's `fit`. `pipe.predict(X_valid)` transforms the validation features using the stored scaling statistics and passes the result to Ridge's `predict`.

```text
fit:      training X ── scaler.fit_transform ── Ridge.fit(y_train)
predict:  new X      ── scaler.transform     ── Ridge.predict ── forecasts
```

The assertion below checks that this produces the same predictions as our manual calculation. `pipe['scale']` and `pipe['model']` let us inspect the fitted components. Calling `predict` should leave both the scaler's means and the model's coefficients unchanged.

Every intermediate pipeline step must transform data so the next step has inputs to work with. Our last step is Ridge, so the complete pipeline predicts returns; it does not expose a final `transform` operation. Keeping all these steps in one estimator will also let cross-validation refit the preprocessing together with the model.

````python
pipe = Pipeline([('scale', StandardScaler()), ('model', Ridge(alpha=100))])
assert pipe.fit(X_train, y_train) is pipe
pipeline_prediction = pipe.predict(X_valid)
np.testing.assert_allclose(pipeline_prediction, manual_prediction)

# named_steps and indexing expose the fitted components for inspection.
learned_mean = pipe.named_steps['scale'].mean_.copy()
learned_coef = pipe['model'].coef_.copy()
_ = pipe.predict(X_valid.iloc[:5])
np.testing.assert_array_equal(pipe['scale'].mean_, learned_mean)
np.testing.assert_array_equal(pipe['model'].coef_, learned_coef)
display(valid[['date', 'symbol']].head().assign(prediction=pipeline_prediction[:5]))
print('Predict is available:', hasattr(pipe, 'predict'))
print('Transform is available:', hasattr(pipe, 'transform'))
````

### Change hyperparameters without overwriting the fitted model

Ridge's `alpha` is a **hyperparameter**: we choose its value before fitting, and it controls the penalty on coefficient sizes. The coefficients in `coef_` are **learned parameters**. Changing `alpha` from 100 to 1000 does not recalculate them; we must call `fit` again to learn coefficients under the new penalty.

`clone(pipe)` creates a new, unfitted pipeline with the same hyperparameters, letting us fit another model while preserving the original. The new scaler has no `mean_`, and the new Ridge has no `coef_`. By contrast, `another = pipe` would only give us a second name for the existing object; fitting it would overwrite the original object's learned state too.

The hyperparameter name `model__alpha` tells `set_params` to find the step called `model` and change its `alpha`. The double underscore separates the step name from the hyperparameter name. `get_params(deep=True)` lists these nested hyperparameters. The code below clones the pipeline, changes `alpha`, and refits the clone.

````python
display(pd.Series({key: value for key, value in pipe.get_params(deep=True).items()
                  if key in ['scale__with_mean', 'model__alpha', 'model__fit_intercept']}))
fresh = clone(pipe).set_params(model__alpha=1000)
assert not hasattr(fresh['scale'], 'mean_')
assert not hasattr(fresh['model'], 'coef_')
assert pipe.get_params()['model__alpha'] == 100
print('Original fitted / clone fitted:', hasattr(pipe['model'], 'coef_'), hasattr(fresh['model'], 'coef_'))
fresh.fit(X_train, y_train)
assert fresh['model'].alpha == 1000
print('Clone fitted after fit:', hasattr(fresh['model'], 'coef_'))
````

## 6. Write a transformer for extreme feature values

Some stocks in the extract have very large trailing returns or financial ratios. A few such observations can dominate a regression fit. One possible treatment is to cap each feature at its training-set 1st and 99th percentiles. Values inside the bounds remain unchanged; values outside are moved to the nearest bound. We keep the rows rather than dropping the observations.

The percentile choices, `lower=0.01` and `upper=0.99`, are hyperparameters chosen before fitting. The actual cutoff values depend on the training sample and are learned attributes, `lower_bounds_` and `upper_bounds_`. `transform` must reuse these bounds. Recalculating them on a prediction batch would make a stock's forecast depend on which other rows happened to be predicted alongside it.

To make this work like a built-in sklearn transformer, implement three methods:

- `__init__` stores the hyperparameters unchanged. It does not inspect data or calculate bounds.
- `fit` checks the input, learns one pair of bounds per column, and returns `self`.
- `transform` checks that fitting has happened and applies those stored bounds without changing the number or order of rows.

The inherited classes provide the remaining methods. `BaseEstimator` supplies `get_params` and `set_params`. A **mixin** is a small base class that adds reusable behavior: `TransformerMixin` supplies `fit_transform`, and `OneToOneFeatureMixin` supplies output names because each input column still corresponds to one output column. These mixins appear before `BaseEstimator` in the inheritance list.

`validate_data(reset=True)` records the feature count and names during fit. With `reset=False`, it checks new inputs against that fitted schema, catching errors such as reordered named columns. Missing values remain missing, and infinities are rejected. The table and histogram below show how much clipping changes the validation data. An extreme source value still needs investigation; clipping is a modeling choice, not a correction to the database.

````python
from sklearn.base import BaseEstimator, TransformerMixin, OneToOneFeatureMixin
from sklearn.utils.validation import check_is_fitted, validate_data

class QuantileClipper(TransformerMixin, OneToOneFeatureMixin, BaseEstimator):
    """Learn per-column tail cutoffs on training rows; preserve missing values."""
    def __init__(self, lower=0.01, upper=0.99):
        self.lower = lower
        self.upper = upper

    def fit(self, X, y=None):
        if not 0 <= self.lower < self.upper <= 1:
            raise ValueError("Require 0 <= lower < upper <= 1")
        a = validate_data(self, X, reset=True, dtype=float, ensure_all_finite="allow-nan")
        # All-missing columns are left unbounded and handled by the imputer.
        self.lower_bounds_ = np.full(a.shape[1], -np.inf)
        self.upper_bounds_ = np.full(a.shape[1], np.inf)
        observed = np.isfinite(a).any(axis=0)
        self.lower_bounds_[observed] = np.nanquantile(a[:, observed], self.lower, axis=0)
        self.upper_bounds_[observed] = np.nanquantile(a[:, observed], self.upper, axis=0)
        return self

    def transform(self, X):
        check_is_fitted(self, "lower_bounds_")
        a = validate_data(self, X, reset=False, dtype=float, ensure_all_finite="allow-nan")
        return np.clip(a, self.lower_bounds_, self.upper_bounds_)
````

````python
clipper = QuantileClipper(lower=0.01, upper=0.99).set_output(transform='pandas')
clipped_train = clipper.fit_transform(X_train)
bounds = clipper.upper_bounds_.copy()
clipped_valid = clipper.transform(X_valid)
np.testing.assert_array_equal(clipper.upper_bounds_, bounds)
assert clipped_valid.index.equals(X_valid.index)
assert clipped_valid.columns.equals(X_valid.columns)
display(pd.DataFrame({'lower_bound': clipper.lower_bounds_, 'upper_bound': bounds,
                      'validation_fraction_clipped': (clipped_valid != X_valid).mean()},
                     index=small_features))

fig, ax = plt.subplots(figsize=(9, 3))
X_valid.momentum_63.hist(bins=50, alpha=0.55, ax=ax, label='Observed validation feature')
clipped_valid.momentum_63.hist(bins=50, alpha=0.55, ax=ax, label='Training-bound clipping')
ax.set(xlabel='63-session trailing return', ylabel='Stock-date rows')
ax.legend()
plt.show()
````

We can now add `QuantileClipper` as a pipeline step. The name `clip` lets us configure its hyperparameters through names such as `clip__lower`.

````python
clipped_pipe = Pipeline([('clip', QuantileClipper()),
                         ('scale', StandardScaler()), ('model', Ridge(alpha=100))])
clipped_pipe.fit(X_train, y_train)
assert np.isfinite(clipped_pipe.predict(X_valid)).all()
print('Nested transformer hyperparameter:', clipped_pipe.get_params()['clip__lower'])
np.testing.assert_array_equal(clipped_pipe[:-1].get_feature_names_out(), small_features)
````
