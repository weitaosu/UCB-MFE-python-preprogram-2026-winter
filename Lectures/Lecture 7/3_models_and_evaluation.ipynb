# 3. Fit a return model and evaluate its predictions

We want to choose a model that helps us select stocks. A model learns from historical examples, but fitting old returns well does not guarantee useful forecasts. We need to test its predictions on later observations it did not learn from.

We'll work through one fitted model first, then compare several models using the same data and evaluation periods. The classroom experiment uses **2010–2014 for model development** and saves **2015 for the portfolio backtest in Lecture 8**. HW4 has its own later test period.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display
import sys
from sklearn.base import clone
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
%matplotlib inline

course_root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').is_file())
modeling_dir = course_root / 'Lectures/modeling'
if str(modeling_dir) not in sys.path:
    sys.path.insert(0, str(modeling_dir))
from course_data import prepare_panel, make_pipeline, prediction_metrics, PRICE_FEATURES, FEATURES
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
```

## 1. What does the model learn from?

A training example pairs **what we knew about a stock** with **the return that followed**. In sklearn, the inputs are called `X` and the return we want to predict is called `y`, or the **target**.

For example, after the close on **December 3, 2012**, AAPL's trailing 63-session return was about **−11.27%** and its annualized trailing volatility was **30.32%**. Those are two possible input features. We want to use information like this to forecast a subsequent holding return.

Our strategy would enter at the next close, December 4, and exit 90 calendar days later, March 4, 2013. AAPL's adjusted-price return over that holding period was **−26.72%**. That is this observation's target. It is a different return from the trailing −11.27% used as an input.

| Part of the example | Meaning | Column in our table |
|---|---|---|
| Observation date | When we calculate the inputs and make a forecast | `date` (called the signal date in earlier notebooks) |
| Features, `X` | Price and financial information available on that date | Columns listed in `FEATURES` |
| Target, `y` | The subsequent 90-calendar-day holding return | `target` |
| Return completion date | When the exit price makes that target observable | `label_end` |

`prepare_panel` calculates these columns from the class database. The display below shows a few features and the trading dates for that AAPL example. `X` contains only the chosen input columns; `y` contains the corresponding returns. The dates help us select rows but are not model inputs.

```python
panel = prepare_panel()
# Keep development outcomes before the fit used for the later 2015 experiment.
dev = panel.loc[panel.eligible & panel.date.ge('2010-01-01')
                & panel.date.lt('2015-01-01') & panel.label_end.lt('2014-12-31')].reset_index(drop=True)
X, y = dev[FEATURES], dev.target
worked_example = dev.loc[dev.symbol.eq('AAPL') & dev.date.eq('2012-12-03')]
display(worked_example[['date', 'symbol', 'momentum_63', 'volatility_63', 'net_margin',
                        'entry_date', 'label_end', 'target']].T)
print(f'{len(dev):,} examples; {len(FEATURES)} input columns; {dev.date.nunique()} observation dates')
```

## 2. Fit on completed examples, then predict a later year

**Fitting** means learning a relationship between the input features and the target returns. A linear model learns weights for the input columns and combines them to estimate a return. During training, it compares those estimates with the actual returns and chooses weights that reduce the errors. Ridge also penalizes large weights to reduce sensitivity to noisy inputs. The penalty strength `alpha` is a **hyperparameter** chosen before fitting; the weights themselves are **learned parameters** estimated during fitting.

We call `model.fit(X_train, y_train)` to learn those weights. Once fitted, `model.predict(X_new)` applies the learned relationship to new feature rows. We do not give `predict` their actual returns. Those returns are what we are trying to forecast.

To check whether this works, we will fit a model before the first forecast of **2013**, then predict the 2013 observations. This later evaluation period is our **validation set**. Evaluating only on the rows used for fitting would reward the model for reproducing answers it had already learned from.

There is an important restriction on the training set: **every return passed to `fit` must have finished before the model is fitted**. For our December 3 AAPL example, the inputs were known in December, but the −26.72% target was only known on March 4. A model fitted at the start of January cannot use that example yet. Although today's historical database contains the March outcome, handing it to a January model would give that model future information.

The code below therefore requires both `date < first_prediction_date` and `label_end < first_prediction_date` for training. It then performs three separate operations:

1. **Fit:** give the model the earlier input rows and their completed returns.
2. **Predict:** give it the 2013 input rows, without their returns.
3. **Evaluate:** compare its predictions with the returns that subsequently occurred.

We can do step 3 today because the history is available. In an actual 2013 run, we would have had to wait for each holding period to finish. The output places each forecast beside its actual return; a positive `error` means the model predicted too high.

```python
validation_mask = dev.date.between('2013-01-01', '2013-12-31')
first_prediction_date = dev.loc[validation_mask, 'date'].min()
training_mask = dev.date.lt(first_prediction_date) & dev.label_end.lt(first_prediction_date)
training = dev.loc[training_mask]
validation = dev.loc[validation_mask]

model = make_pipeline(Ridge(alpha=100))
model.fit(training[FEATURES], training.target)
forecast = model.predict(validation[FEATURES])

comparison = validation[['date', 'symbol', 'target']].rename(columns={'target': 'actual_return'}).copy()
comparison['prediction'] = forecast
comparison['error'] = comparison.prediction - comparison.actual_return
assert training.label_end.max() < first_prediction_date
assert not training.index.intersection(worked_example.index).size
print('First prediction date:', first_prediction_date.date())
print('Training examples:', len(training), '; validation examples:', len(validation))
print('December AAPL example included in this fit:', bool(training.index.isin(worked_example.index).any()))
display(comparison[['date', 'symbol', 'prediction', 'actual_return', 'error']].head(8))
```

## 3. Repeat the comparison in several years

One year may flatter a model that happened to suit that market. Repeat the same fit–predict–evaluate process for 2012, 2013, and 2014. Each repetition is called a **validation fold**.

| Fold | Training examples | Observations to predict and evaluate |
|---|---|---|
| 1 | From 2010 onward, with returns completed before the first 2012 forecast | 2012 |
| 2 | From 2010 onward, with returns completed before the first 2013 forecast | 2013 |
| 3 | From 2010 onward, with returns completed before the first 2014 forecast | 2014 |

Each fold fits a new model once and holds it fixed for that year's predictions. Later folds can train on more completed history. We keep every stock on a validation date in validation, since that day's forecasts should all come from the model fitted before the year began. A random split could instead train on later observations and evaluate earlier ones, which would not reproduce that sequence.

`date_folds` repeats the two filters from our 2013 example. Removing training rows whose future returns are unfinished is often called **purging**. The function returns pairs of integer row positions: `tr` selects training rows, and `va` selects validation rows. sklearn uses these positions to take matching rows from `X` and `y`.

The table and chart show the resulting periods. Training observation dates stop before validation because the recent 90-day returns are still unfinished. The 2014 validation period also ends early: later observations cannot be scored without outcomes beyond our development-data cutoff. This limits prediction-error measurement; it is not a rule for excluding stocks from a live portfolio.

```python
def date_folds(frame, validation_years=(2012, 2013, 2014), train_start='2010-01-01'):
    """Choose earlier completed examples for each later validation year."""
    splits = []
    for year in validation_years:
        validation_mask = frame.date.between(f'{year}-01-01', f'{year}-12-31') & frame.target.notna()
        if not validation_mask.any():
            raise ValueError(f'No validation rows for {year}')
        first_prediction_date = frame.loc[validation_mask, 'date'].min()
        training_mask = (frame.date.ge(train_start) & frame.date.lt(first_prediction_date)
                         & frame.label_end.lt(first_prediction_date) & frame.target.notna())
        tr = np.flatnonzero(training_mask)
        va = np.flatnonzero(validation_mask)
        if not len(tr):
            raise ValueError(f'No completed training returns before {year}')
        assert frame.iloc[tr].label_end.max() < frame.iloc[va].date.min()
        assert not set(frame.iloc[tr].date).intersection(frame.iloc[va].date)
        splits.append((tr, va))
    return splits
```

```python
validation_years = (2012, 2013, 2014)
folds = date_folds(dev, validation_years=validation_years)
fold_rows = []
for fold, (tr, va) in enumerate(folds, start=1):
    fold_rows.append({'fold': fold, 'validation_year': validation_years[fold-1],
                      'training_rows': len(tr), 'validation_rows': len(va),
                      'last_training_observation': dev.iloc[tr].date.max(),
                      'last_completed_training_return': dev.iloc[tr].label_end.max(),
                      'first_validation_observation': dev.iloc[va].date.min(),
                      'last_validation_observation': dev.iloc[va].date.max()})
display(pd.DataFrame(fold_rows))
fig, ax = plt.subplots(figsize=(10, 3))
for i, (tr, va) in enumerate(folds):
    ax.plot([dev.iloc[tr].date.min(), dev.iloc[tr].date.max()], [i, i], lw=10, color='steelblue')
    ax.plot([dev.iloc[va].date.min(), dev.iloc[va].date.max()], [i, i], lw=10, color='darkorange')
ax.set(yticks=range(len(folds)), yticklabels=[f'{year} validation' for year in validation_years],
       title='Observation dates: blue = training; orange = validation')
plt.tight_layout()
```

## 4. Which model makes smaller return errors?

Now repeat these same experiments with five models. Keep their inputs and validation periods the same so the results compare the models rather than different datasets.

| Model | What it fits | Why try it? |
|---|---|---|
| Training mean | One average of the training returns | Establishes what we can predict without using the stock features |
| OLS | A weight for each input and an intercept | Tests whether a linear combination of the features is useful |
| Ridge | Linear weights with a squared-weight penalty | Limits large weights, especially with noisy or correlated features |
| Lasso | Linear weights with an absolute-weight penalty | Can set some weights to zero and use fewer inputs |
| Boosted trees | A sequence of trees that corrects earlier errors | Allows thresholds and interactions, such as momentum behaving differently at high volatility |

`clone(estimator)` creates an unfitted copy for each experiment. Fitting the whole pipeline also relearns clipping, missing-value replacement, sector encoding, and scaling using only that fold's training rows. We disable the tree model's random internal early-stopping split and use our own later-year results to evaluate it.

Read the error columns as follows:

- **MAE:** the average absolute prediction error. A forecast of 5% followed by an actual −5% return has an absolute error of 10 percentage points.
- **RMSE:** square the errors, average them, then take the square root. This gives larger misses more influence than MAE. Both metrics use return decimals: `0.10` means ten percentage points.
- **R²:** compares squared errors with predicting the validation sample's mean return for every row. Negative R² means the model's squared errors are larger. That sample mean is known afterward; the training-mean model is a separate forecast we could have made at the time.

Compare all three years before averaging. A good average can hide a model that performs well in one year and poorly in the others.

```python
candidates = {
    'training mean': make_pipeline(DummyRegressor(strategy='mean')),
    'OLS': make_pipeline(LinearRegression()),
    'Ridge': make_pipeline(Ridge(alpha=100)),
    'Lasso': make_pipeline(Lasso(alpha=0.001, max_iter=10000)),
    'boosted trees': make_pipeline(HistGradientBoostingRegressor(
        max_iter=60, max_leaf_nodes=7, min_samples_leaf=100,
        l2_regularization=10, early_stopping=False, random_state=42)),
}
results = []
validation_frames = []
for name, estimator in candidates.items():
    for fold, (tr, va) in enumerate(folds, start=1):
        fitted = clone(estimator).fit(X.iloc[tr], y.iloc[tr])
        prediction = fitted.predict(X.iloc[va])
        metrics = prediction_metrics(dev.iloc[va], prediction)
        results.append({'model': name, 'fold': fold, **metrics.to_dict()})
        scored = dev.iloc[va][['date', 'symbol', 'target']].copy()
        scored['prediction'], scored['model'], scored['fold'] = prediction, name, fold
        validation_frames.append(scored)
results = pd.DataFrame(results)
scored_validation = pd.concat(validation_frames, ignore_index=True)
display(results.round(4))
display(results.groupby('model')[['RMSE', 'MAE', 'R2', 'mean_daily_rank_IC']].mean().round(4))
```

## 5. Does the model put the better stocks near the top?

In this class extract, the training-mean model has the lowest average RMSE. But it gives every stock the same forecast, so it cannot tell us which stocks to buy. Return accuracy and useful stock ranking are related questions with different answers.

For each observation date, compare two lists: the stocks ordered by predicted return, and the same stocks ordered by the return they later earned. **Spearman rank correlation** measures how closely those orderings agree. In stock-selection research, this is often called the **information coefficient (IC)**.

A positive IC means higher forecasts generally went with higher realized returns; a negative IC means the ranking tended to run the wrong way. An IC of 0.2 is a correlation, not a 20% investment return. Calculate it separately for each date, so we measure stock selection within the day rather than mixing it with differences between market periods.

The first table below shows Ridge's predicted and actual ranks on one date. The chart then tracks monthly averages of the daily IC. Finally, we divide each day's stocks into five forecast groups: bucket 1 has the lowest predictions and bucket 5 the highest. Compare their average actual returns to see whether higher forecasts were followed by better outcomes.

```python
ridge_scores = scored_validation.loc[scored_validation.model.eq('Ridge')].copy()
example_date = ridge_scores.date.min()
rank_example = ridge_scores.loc[ridge_scores.date.eq(example_date), ['symbol', 'prediction', 'target']].copy()
rank_example['predicted_rank'] = rank_example.prediction.rank(ascending=False, method='min')
rank_example['actual_rank'] = rank_example.target.rank(ascending=False, method='min')
display(rank_example.sort_values('predicted_rank').head(8))

daily_ic = ridge_scores.groupby('date').apply(
    lambda g: g.prediction.corr(g.target, method='spearman'), include_groups=False)
monthly_ic = daily_ic.resample('ME').mean()
monthly_ic.plot(kind='bar', figsize=(12, 3), title='Ridge: monthly average of daily ranking correlation')
plt.axhline(0, color='black', lw=1)
plt.ylabel('Rank correlation')
plt.xticks(range(len(monthly_ic)), monthly_ic.index.strftime('%Y-%m'), rotation=90)
plt.tight_layout()

ridge_scores = ridge_scores.sort_values(['date', 'prediction', 'symbol'])
ridge_scores['bucket'] = (ridge_scores.groupby('date').prediction.rank(method='first', pct=True)
                          .mul(5).apply(np.ceil).clip(1, 5))
display(ridge_scores.groupby('bucket').target.agg(['mean', 'std', 'count']))
```

These buckets average hypothetical holding returns, without deciding how much money could actually be invested. They therefore do not describe a portfolio's return. Lecture 8 adds cash, position sizes, fees, and overlapping purchases.

The observation count also overstates the amount of independent evidence. Trades beginning on adjacent days share almost the same 90-day price path. A single rally can improve many of their outcomes together, so check the separate years and periods rather than treating every row as an independent confirmation.

## 6. Tune Ridge's alpha hyperparameter

The hyperparameter `alpha` controls Ridge's penalty on its learned weights. Larger values shrink weights more strongly. We will compare `1`, `100`, and `1000`, choosing the value with the smallest average validation RMSE.

`GridSearchCV` runs the fitting loop for us. It creates a fresh pipeline for each penalty and fold, fits the training rows, predicts the validation rows, and records the error. Three penalties and three folds require nine validation fits. The name `model__alpha` tells it to change `alpha` in the pipeline step called `model`.

The scorer is `neg_root_mean_squared_error` because sklearn searches for the largest score: −0.12 is better than −0.15, corresponding to a smaller RMSE. The output below converts these back to positive errors. In sklearn's raw results, `mean_test_score` means the validation folds we supplied, not the reserved 2015 test period.

With `refit=True`, the search then fits a model with the chosen hyperparameter value on all the development rows and stores that model in `best_estimator_`. It is ready for the later test because those training outcomes were completed before that test began. Choosing a hyperparameter value from these folds can favor one that was lucky, which is why we still need the separate 2015 evaluation.

```python
search = GridSearchCV(make_pipeline(Ridge()),
                      param_grid={'model__alpha': [1, 100, 1000]},
                      cv=folds, scoring='neg_root_mean_squared_error',
                      n_jobs=1, return_train_score=True, error_score='raise')
search.fit(X, y)
cv_table = pd.DataFrame(search.cv_results_)
search_summary = pd.DataFrame({
    'alpha': cv_table.param_model__alpha,
    'mean_training_RMSE': -cv_table.mean_train_score,
    'mean_validation_RMSE': -cv_table.mean_test_score,
    'validation_RMSE_std': cv_table.std_test_score,
    'rank': cv_table.rank_test_score,
})
display(search_summary)
print('Selected hyperparameters:', search.best_params_)
print('Best mean validation RMSE:', -search.best_score_)
```

## 7. Did adding financial information help?

We have used price features, financial ratios, filing age, and sector indicators together. To test whether those extra inputs helped Ridge, compare that full model with one using only the price features. Removing a group of inputs for a comparison is called a **feature ablation**.

Fit both versions again at the same selected penalty, on the same three folds. Compare the paired RMSE and IC results within each year. A lower RMSE for price-only means that adding the other inputs worsened return accuracy with these hyperparameters. It does not identify a particular ratio as the cause, because we removed ratios, filing age, and sectors together.

This holds the hyperparameter `alpha` fixed to examine the input change. Separately tuning each feature set would be another experiment, asking how well each set could perform after its own model selection.

```python
selected_alpha = search.best_params_['model__alpha']
ablation_candidates = {
    'full features': make_pipeline(Ridge(alpha=selected_alpha)),
    'price only': make_pipeline(Ridge(alpha=selected_alpha),
                               numeric_features=PRICE_FEATURES, include_sector=False),
}
ablation_rows = []
for feature_set, estimator in ablation_candidates.items():
    for fold, (tr, va) in enumerate(folds, start=1):
        fitted = clone(estimator).fit(X.iloc[tr], y.iloc[tr])
        metrics = prediction_metrics(dev.iloc[va], fitted.predict(X.iloc[va]))
        ablation_rows.append({'features': feature_set, 'alpha': selected_alpha,
                              'fold': fold, **metrics.to_dict()})
display(pd.DataFrame(ablation_rows).sort_values(['fold', 'features']))
```

### Optional: predict the probability of a positive return

So far the target has been a numeric return. We can instead ask whether that return will be positive, replacing `y` with the zero/one outcome `y > 0`. This turns the task from regression into classification.

`LogisticRegression.fit` learns from the same features and these binary outcomes. `predict_proba` returns a probability for each class; `[:, 1]` selects the positive-return probability here. A threshold of 0.5 produces the decisions in the confusion matrix.

**Precision** measures how many predicted positives were actually positive; **recall** measures how many actual positives we found. **ROC AUC** measures whether positives receive higher probabilities than nonpositives across thresholds. **Brier score** is the mean squared error between probabilities and zero/one outcomes, where lower is better.

Compare accuracy with the fraction of positive outcomes: when most stocks rise, always predicting a positive return may already achieve high accuracy. Also consider the sizes of gains and losses. Getting many small winners right does not guarantee a profitable strategy if the missed losses are large.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score, brier_score_loss, confusion_matrix
tr, va = folds[-1]
classifier = make_pipeline(LogisticRegression(C=0.1, max_iter=1000))
classifier.fit(X.iloc[tr], y.iloc[tr].gt(0))
probability = classifier.predict_proba(X.iloc[va])[:, 1]
event = y.iloc[va].gt(0)
decision = probability >= 0.5
display(pd.Series({'positive_fraction': event.mean(), 'accuracy': accuracy_score(event, decision),
                   'precision': precision_score(event, decision, zero_division=0),
                   'recall': recall_score(event, decision, zero_division=0),
                   'ROC_AUC': roc_auc_score(event, probability),
                   'Brier': brier_score_loss(event, probability)}))
display(pd.DataFrame(confusion_matrix(event, decision), index=['actual nonpositive', 'actual positive'],
                     columns=['predict nonpositive', 'predict positive']))
```
